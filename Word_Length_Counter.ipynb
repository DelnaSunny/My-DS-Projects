```python
from collections import defaultdict
import re
```

```python
data = []
with open("/content/Sowpods.txt", 'r') as f:
    for line in f:
        data.append(line.strip())
```

```python
def count_word_lengths(word_list):
    length_counts = defaultdict(int)

    for word in word_list:
        length = len(word)
        length_counts[length] += 1

    return dict(sorted(length_counts.items()))


word_length_counts = count_word_lengths(data)
print("Word Length Counts:")
for length, count in word_length_counts.items():
    print(f"Length{length}:{count} word(s)")
```
